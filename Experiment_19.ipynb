**Aim: Write a program to implement Longest Common Subsequence (LCS) using Dynamic Programming.**

**Algorithm:**

Longest Common Subsequence (LCS) using Dynamic Programming

Step 1: Start

Step 2: Input two strings X and Y

Step 3: Let m = length of X, n = length of Y

Step 4: Create a DP table dp of size (m+1) × (n+1)

Step 5: Initialize:

        dp[i][0] = 0 for all i
        dp[0][j] = 0 for all j

Step 6: Fill DP table:

        for i = 1 to m
            for j = 1 to n
                if X[i-1] == Y[j-1]
                    dp[i][j] = 1 + dp[i-1][j-1]
                else
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])

Step 7: LCS length = dp[m][n]

Step 8: Backtrack to find LCS sequence:

        Start from dp[m][n]
        If characters match → include in LCS
        Else move to max(dp[i-1][j], dp[i][j-1])

Step 9: Print LCS length and sequence

Step 10: Stop

```python
def lcs_dp(X, Y):
    m, n = len(X), len(Y)

    dp = [[0]*(n+1) for _ in range(m+1)]

    # Fill DP Table
    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i-1] == Y[j-1]:
                dp[i][j] = 1 + dp[i-1][j-1]
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])

    return dp


def get_lcs(dp, X, Y):
    i, j = len(X), len(Y)
    lcs_seq = []

    while i > 0 and j > 0:
        if X[i-1] == Y[j-1]:
            lcs_seq.append(X[i-1])
            i -= 1
            j -= 1
        elif dp[i-1][j] > dp[i][j-1]:
            i -= 1
        else:
            j -= 1

    return ''.join(reversed(lcs_seq))


while True:

    print("\n===== LCS MENU =====")
    print("1. Find LCS")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        X = input("Enter first string: ")
        Y = input("Enter second string: ")

        dp = lcs_dp(X, Y)

        lcs_length = dp[len(X)][len(Y)]
        lcs_sequence = get_lcs(dp, X, Y)

        print("\nLCS Length:", lcs_length)
        print("LCS Sequence:", lcs_sequence)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```
