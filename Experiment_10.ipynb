#**Aim: Implement Algorithm to Compute Maximum Weight Maximal Independent Set in a Graph**

**Algorithm**

1. Start
2. Read vertices and their weights
3. Read edges
4. Build adjacency list
5. Initialize:
   * best_weight = 0
   * best_set = empty
   * operation_count = 0
6. Define recursive function backtrack(current_set, remaining_vertices)
7. For each vertex:
   * Check if it is independent with current_set
   * Increase operation count
   * If valid:
     * Add to current_set
     * Recur for remaining vertices
     * Remove (backtrack)
8. Update best_weight if current_set weight > best_weight
9. After exploring all possibilities, print best_set and best_weight
10. Stop

```python
def is_independent(vertex, current_set, graph, op_count):
    for v in current_set:
        op_count[0] += 1
        if vertex in graph[v]:
            return False
    return True


def backtrack(vertices, graph, weights, index,
              current_set, current_weight, best, op_count):
    op_count[0] += 1

    if index == len(vertices):
        if current_weight > best[0]:
            best[0] = current_weight
            best[1] = current_set.copy()
        return

    vertex = vertices[index]

    # Option 1: Include vertex
    if is_independent(vertex, current_set, graph, op_count):
        current_set.append(vertex)
        backtrack(vertices, graph, weights, index + 1,
                  current_set, current_weight + weights[vertex],
                  best, op_count)
        current_set.pop()

    # Option 2: Exclude vertex
    backtrack(vertices, graph, weights, index + 1,
              current_set, current_weight,
              best, op_count)
```

```python
# Main Program
vertices = input("Enter vertex names (space separated): ").split()

weights = {}
print("Enter weights of vertices:")
for v in vertices:
    weights[v] = int(input(f"Weight of {v}: "))

E = int(input("Enter number of edges: "))
edges = []

graph = {v: [] for v in vertices}

print("Enter edges (u v):")
for _ in range(E):
    u, v = input().split()
    graph[u].append(v)
    graph[v].append(u)

while True:
    print("\n--- MAX WEIGHT INDEPENDENT SET MENU ---")
    print("1. Compute Maximum Weight Independent Set")
    print("2. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        best = [0, []]
        op_count = [0]

        backtrack(vertices, graph, weights, 0,
                  [], 0, best, op_count)

        print("\nMaximum Weight Independent Set:", best[1])
        print("Maximum Total Weight:", best[0])
        print("Total Operation Count:", op_count[0])

    elif choice == 2:
        print("\nExperiment Completed Successfully.")
        print("Developed by: Abu Sayem")
        break

    else:
        print("Invalid choice!")
```
