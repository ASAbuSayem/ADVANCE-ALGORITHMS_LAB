**Aim: WAP to implement Floyd–Warshall Algorithm.**

Step 1: Start

Step 2: Input number of vertices (n)

Step 3: Input the weighted adjacency matrix of the graph

Step 4: Initialize distance matrix = adjacency matrix

Step 5: For k = 0 to n-1

            For i = 0 to n-1

                For j = 0 to n-1

                    If dist[i][j] > dist[i][k] + dist[k][j]
                    
                        dist[i][j] = dist[i][k] + dist[k][j]

Step 6: Print the shortest distance matrix

Step 7: Stop

```python
INF = 999999

def floyd_warshall(graph, n):

    dist = []

    for i in range(n):
        row = []
        for j in range(n):
            row.append(graph[i][j])
        dist.append(row)

    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][j] > dist[i][k] + dist[k][j]:
                    dist[i][j] = dist[i][k] + dist[k][j]

    print("\nShortest Distance Matrix:")

    for i in range(n):
        for j in range(n):
            if dist[i][j] == INF:
                print("INF", end=" ")
            else:
                print(dist[i][j], end=" ")
        print()


while True:

    print("\n--- FLOYD WARSHALL ALGORITHM ---")
    print("1. Find Shortest Paths")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        n = int(input("Enter number of vertices: "))

        print("Enter adjacency matrix (use 999999 for INF):")

        graph = []

        for i in range(n):
            row = list(map(int, input().split()))
            graph.append(row)

        floyd_warshall(graph, n)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid choice")
```
