# **Experiment-21**

**Aim: Implement Chinese Remainder Theorem (CRT) to find a number that satisfies given congruences.**

**ALGORITHM :Chinese Remainder Theorem (CRT)**

Step 1: Start

Step 2: Input number of equations (n)

Step 3: Input remainders array a[] and moduli array m[]

Step 4: Compute product M = m1 × m2 × ... × mn

Step 5: For each i from 0 to n-1:

        Mi = M / mi

        Find inverse yi such that:
        (Mi × yi) mod mi = 1

Step 6: Compute result:

        x = Σ (ai × Mi × yi)

Step 7: Final answer:

        x = x mod M

Step 8: Print solution x

Step 9: Stop

**Code:**

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

# Modular inverse using Extended Euclidean Algorithm
def mod_inverse(a, m):
    m0 = m
    x0, x1 = 0, 1

    if m == 1:
        return 0

    while a > 1:
        q = a // m
        a, m = m, a % m
        x0, x1 = x1 - q * x0, x0

    if x1 < 0:
        x1 += m0

    return x1


def chinese_remainder(a, m, n):
    M = 1
    for i in range(n):
        M *= m[i]

    result = 0

    for i in range(n):
        Mi = M // m[i]
        yi = mod_inverse(Mi, m[i])
        result += a[i] * Mi * yi

    return result % M


while True:

    print("\n===== CHINESE REMAINDER THEOREM =====")
    print("1. Solve CRT")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        n = int(input("Enter number of equations: "))

        print("Enter remainders (space separated):")
        a = list(map(int, input().split()))

        print("Enter moduli (space separated):")
        m = list(map(int, input().split()))

        if len(a) != n or len(m) != n:
            print("Invalid input!")
            continue

        # Check pairwise coprime
        valid = True
        for i in range(n):
            for j in range(i+1, n):
                if gcd(m[i], m[j]) != 1:
                    valid = False

        if not valid:
            print("Moduli are not pairwise coprime. CRT not applicable.")
            continue

        result = chinese_remainder(a, m, n)

        print("\nSolution x =", result)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```
