# **Experiment-24**

**Aim: Write a program to solve the Traveling Salesman Problem (TSP) using an Approximation Algorithm.**

**ALGORITHM: TSP using Approximation (Nearest Neighbor Method)**

Step 1: Start

Step 2: Input number of cities (n)

Step 3: Input distance matrix

Step 4: Select starting city (usually 0)

Step 5: Initialize:

        visited = {starting city}
        current = starting city
        total_cost = 0
        path = [starting city]

Step 6: Repeat until all cities are visited:

        From current city, choose nearest unvisited city

        Add distance to total_cost

        Mark city as visited

        Update current city

        Append city to path

Step 7: Return to starting city

        Add cost

Step 8: Print path and total cost

Step 9: Stop

```python
while True:

    print("\n===== TSP (NEAREST NEIGHBOR APPROXIMATION) =====")
    print("1. Solve TSP")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        n = int(input("Enter number of cities: "))

        print("Enter distance matrix:")

        dist = []
        for i in range(n):
            row = list(map(int, input().split()))
            dist.append(row)

        visited = [False] * n

        start = 0
        current = start
        visited[current] = True

        path = [current]
        total_cost = 0

        for _ in range(n - 1):

            next_city = -1
            min_dist = float('inf')

            for j in range(n):
                if not visited[j] and dist[current][j] < min_dist:
                    min_dist = dist[current][j]
                    next_city = j

            visited[next_city] = True
            path.append(next_city)
            total_cost += min_dist
            current = next_city

        # return to start
        total_cost += dist[current][start]
        path.append(start)

        print("\nPath:", " -> ".join(map(str, path)))
        print("Total Cost:", total_cost)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```

Time Complexity
O(n2)

Space Complexity
O(n)
