**Aim: Write a program to solve the Set Cover Problem using Approximation Algorithm.**

**ALGORITHM: Set Cover Problem (Greedy Approximation)**

Step 1: Start

Step 2: Input universe U (set of all elements)

Step 3: Input collection of subsets S

Step 4: Initialize:

        Covered = empty set
        Selected_Subsets = empty list

Step 5: While Covered ≠ U:

        Select subset Si that covers maximum number of uncovered elements

        Add Si to Selected_Subsets

        Update Covered = Covered ∪ Si

Step 6: Print Selected_Subsets

Step 7: Stop

```python
while True:

    print("\n===== SET COVER (GREEDY APPROXIMATION) =====")
    print("1. Solve Set Cover Problem")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        print("Enter Universe elements (space separated):")
        U = set(map(int, input().split()))

        n = int(input("Enter number of subsets: "))

        subsets = []

        for i in range(n):
            print(f"Enter elements of subset S{i+1}:")
            s = set(map(int, input().split()))
            subsets.append(s)

        covered = set()
        selected = []

        while covered != U:

            best_subset = None
            max_covered = 0

            for s in subsets:
                uncovered = len(s - covered)
                if uncovered > max_covered:
                    max_covered = uncovered
                    best_subset = s

            if best_subset is None:
                print("No solution possible!")
                break

            selected.append(best_subset)
            covered |= best_subset

        print("\nSelected Subsets:")
        for s in selected:
            print(s)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```

Time Complexity
O(n×m)

(n = subsets, m = elements)

Space Complexity
O(n+m)
