**Aim: To implement Edmonds–Karp Algorithm to find the Maximum Flow in a Network.**

**Algorithm: Edmonds–Karp Algorithm**

1. Start

2. Input number of vertices

3. Input vertex names

4. Input number of edges

5. Input edges with capacity (u, v, capacity)

6. Input source (S) and sink (T)

7. Create residual graph

8. Initialize:

      max_flow = 0

9. While BFS finds path from S to T:

      9.1 Find bottleneck capacity (minimum edge capacity in path)

      9.2 Update residual capacities:

            Forward edge -= bottleneck
            
            Reverse edge += bottleneck

      9.3 Add bottleneck to max_flow

10. Repeat until no augmenting path exists

11. Print maximum flow

12. Stop

```python
from collections import defaultdict, deque

class EdmondsKarp:
    def __init__(self):
        self.graph = defaultdict(dict)

    def add_edge(self, u, v, capacity):
        self.graph[u][v] = capacity
        if v not in self.graph or u not in self.graph[v]:
            self.graph[v][u] = 0

    def bfs(self, source, sink, parent):
        visited = set()
        queue = deque([source])
        visited.add(source)

        while queue:
            u = queue.popleft()
            for v in self.graph[u]:
                if v not in visited and self.graph[u][v] > 0:
                    visited.add(v)
                    parent[v] = u
                    queue.append(v)
                    if v == sink:
                        return True
        return False

    def max_flow(self, source, sink):
        parent = {}
        max_flow = 0
        path_number = 1

        while self.bfs(source, sink, parent):
            path_flow = float("inf")
            s = sink
            path = []

            while s != source:
                path_flow = min(path_flow, self.graph[parent[s]][s])
                path.append(s)
                s = parent[s]

            path.append(source)
            path.reverse()

            print(f"\nAugmenting Path {path_number}: {' -> '.join(path)}")
            print("Flow Added:", path_flow)

            max_flow += path_flow

            v = sink
            while v != source:
                u = parent[v]
                self.graph[u][v] -= path_flow
                self.graph[v][u] += path_flow
                v = parent[v]

            path_number += 1

        return max_flow
```

```python
def main():
    ek = EdmondsKarp()

    vertices = input("Enter vertices (space separated): ").split()

    E = int(input("Enter number of edges: "))
    print("Enter edges in format (u v capacity):")

    for _ in range(E):
        u, v, c = input().split()
        ek.add_edge(u, v, int(c))

    source = input("Enter source vertex: ")
    sink = input("Enter sink vertex: ")

    while True:
        print("\n--- EDMONDS-KARP MENU ---")
        print("1. Compute Maximum Flow")
        print("2. Exit")

        choice = input("Enter choice: ")

        if choice == '1':
            maxflow = ek.max_flow(source, sink)
            print("\nMaximum Flow from", source, "to", sink, "=", maxflow)

        elif choice == '2':
            print("Program Completed")
            print("Developed by: Abu Sayem")
            break

        else:
            print("Invalid choice!")

if __name__ == "__main__":
    main()
```

- The residual graph is updated after every augmentation.

- The algorithm terminates when no path exists from source to sink in the residual graph.

- Edmonds–Karp is a specific implementation of the Ford–Fulkerson method using BFS.

- It guarantees polynomial time complexity unlike the generic Ford–Fulkerson method.
