**Aim: Write a program to implement Matrix Chain Multiplication using Dynamic Programming.**

**ALGORITHM**

Matrix Chain Multiplication (Dynamic Programming)

Step 1: Start

Step 2: Input number of matrices (n)

Step 3: Input dimensions array p[]

        (size = n+1, where matrix Ai is p[i-1] × p[i])

Step 4: Create DP table m[n][n]

Step 5: Initialize:

        m[i][i] = 0 for all i

Step 6: For chain length L = 2 to n:

            for i = 0 to n-L:
                j = i + L - 1
                set m[i][j] = infinity
                for k = i to j-1:
                    cost = m[i][k] + m[k+1][j] + p[i]*p[k+1]*p[j+1]
                    if cost < m[i][j]:
                        m[i][j] = cost
                        store k as split point

Step 7: Minimum cost = m[0][n-1]

Step 8: Print optimal parenthesization using split table

Step 9: Stop

```python
def matrix_chain(p):
    n = len(p) - 1

    m = [[0]*n for _ in range(n)]
    split = [[0]*n for _ in range(n)]

    for L in range(2, n+1):
        for i in range(n-L+1):
            j = i + L - 1
            m[i][j] = float('inf')

            for k in range(i, j):
                cost = m[i][k] + m[k+1][j] + p[i]*p[k+1]*p[j+1]

                if cost < m[i][j]:
                    m[i][j] = cost
                    split[i][j] = k

    return m, split


def print_order(split, i, j):
    if i == j:
        return f"A{i+1}"
    else:
        k = split[i][j]
        left = print_order(split, i, k)
        right = print_order(split, k+1, j)
        return f"({left} × {right})"


while True:

    print("\n===== MATRIX CHAIN MULTIPLICATION =====")
    print("1. Find Minimum Multiplication Cost")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        n = int(input("Enter number of matrices: "))

        print("Enter dimensions (space separated):")
        print("Example: For A1(10x30), A2(30x5) → input: 10 30 5")

        p = list(map(int, input().split()))

        if len(p) != n + 1:
            print("Invalid input! Dimensions must be n+1 values.")
            continue

        m, split = matrix_chain(p)

        print("\nMinimum Multiplication Cost:", m[0][n-1])

        order = print_order(split, 0, n-1)
        print("Optimal Parenthesization:", order)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```
