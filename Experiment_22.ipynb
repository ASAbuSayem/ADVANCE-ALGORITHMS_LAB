**Aim: Write a program to implement Fast Fourier Transform (FFT) using Divide and Conquer approach.**

**ALGORITHM: Fast Fourier Transform (FFT)**

Step 1: Start

Step 2: Input number of elements (n)

        (n must be power of 2)

Step 3: Input sequence x[n]

Step 4: If n == 1:

            return x

Step 5: Divide sequence into:

            Even indexed elements
            Odd indexed elements

Step 6: Recursively compute:

            FFT(even)
            FFT(odd)

Step 7: Combine results:

        For k = 0 to n/2 - 1
            t = Wn^k × FFT(odd)[k]
            X[k] = FFT(even)[k] + t
            X[k + n/2] = FFT(even)[k] - t

        where Wn = e^(-2πi/n)

Step 8: Return result array X

Step 9: Stop

```python
import cmath
import math

def fft(x):
    n = len(x)

    if n == 1:
        return x

    even = fft(x[0::2])
    odd = fft(x[1::2])

    T = [cmath.exp(-2j * math.pi * k / n) * odd[k] for k in range(n // 2)]

    return [even[k] + T[k] for k in range(n // 2)] + \
           [even[k] - T[k] for k in range(n // 2)]


while True:

    print("\n===== FAST FOURIER TRANSFORM =====")
    print("1. Compute FFT")
    print("2. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        n = int(input("Enter number of elements (power of 2): "))

        if (n & (n - 1)) != 0:
            print("Error: n must be power of 2")
            continue

        print("Enter elements (space separated):")
        arr = list(map(float, input().split()))

        if len(arr) != n:
            print("Invalid input length!")
            continue

        result = fft(arr)

        print("\nFFT Result:")
        for val in result:
            print(val)

    elif choice == 2:
        print("Program Exited")
        break

    else:
        print("Invalid Choice")
```

Time Complexity
O(nlogn)

Space Complexity
O(n)
