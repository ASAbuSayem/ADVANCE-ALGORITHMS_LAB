# **Aim: Sort Graph Vartices Using Topological Sorting in Directed a Cyclic Graph**

**Algorithm Steps:**

1. Start
2. Read vertex names
3. Read directed edges
4. Build adjacency list
5. Initialize:
   * visited set
   * rec_stack set (cycle detection)
   * stack (topological order)
   * operation_count = 0
6. For each vertex:
   * If not visited, perform DFS
7. In DFS:
   * Mark vertex visited
   * Add vertex to recursion stack
   * For each adjacent vertex:
     * Increase operation count
     * If not visited → DFS
     * If in recursion stack → cycle detected
8. After exploring all neighbors:
   * Remove vertex from recursion stack
   * Push vertex into stack
9. If cycle detected:
   * Print “Topological Sort not possible”
10. Else:
   * Reverse stack → Topological Order
11. Display result and operation count
12. Stop

```python
def dfs(vertex, graph, visited, rec_stack, stack, op_count):
    visited.add(vertex)
    rec_stack.add(vertex)

    for neighbor in graph.get(vertex, []):
        op_count[0] += 1  # edge check count

        if neighbor not in visited:
            if dfs(neighbor, graph, visited, rec_stack, stack, op_count):
                return True
        elif neighbor in rec_stack:
            return True  # Cycle detected

    rec_stack.remove(vertex)
    stack.append(vertex)
    return False


def topological_sort(vertices, edges):
    graph = {v: [] for v in vertices}
    for u, v in edges:
        graph[u].append(v)

    visited = set()
    rec_stack = set()
    stack = []
    op_count = [0]

    for v in vertices:
        if v not in visited:
            if dfs(v, graph, visited, rec_stack, stack, op_count):
                return None, op_count[0]

    stack.reverse()
    return stack, op_count[0]
```

```python
#main program
vertices = input("Enter vertex names (space separated, e.g. a b c d): ").split()
edges = []

E = int(input("Enter number of directed edges: "))
print("Enter edges in format: source destination")
print("Example: a b (means a -> b)")

for _ in range(E):
    u, v = input().split()
    edges.append((u, v))

while True:
    print("\n--- TOPOLOGICAL SORT MENU ---")
    print("1. Perform Topological Sort (DFS Based)")
    print("2. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        order, operations = topological_sort(vertices, edges)

        if order is None:
            print("\nCycle Detected!")
            print("Topological Sorting is NOT possible.")
        else:
            print("\nTopological Order:")
            print(" -> ".join(order))

        print("Total Operation Count:", operations)

    elif choice == 2:
        print("\nThank You!")
        print("Experiment Completed Successfully.")
        print("Developed by: Abu Sayem")
        break

    else:
        print("Invalid choice! Try again.")
```
